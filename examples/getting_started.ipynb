### Purpose 
- examples of how to get started with latlng-to-state 
- import state data files 
- check a few points and see what state they belong too

#### Imports

```python
# folium for plotting geo
import folium

# access package data
from latlng_to_state import datasets
# look up geometry by point
from latlng_to_state.core.lookup import lookup_geometry_by_point
```

#### Data

```python
state = datasets.load_us_state_geometry()
```

```python
state.head()
```

Note there is a geometry type in the state dataframe 

```python
state.dtypes
```

#### Find what state a point belongs too 

```python
# point in TX
print(
    lookup_geometry_by_point(
        lat=31.203405,
        lng=-99.592150,
        gdf=state,
        return_col='state',
    )
)
```

```python
# point in WA
print(
    lookup_geometry_by_point(
        lat=47.605237,
        lng=-122.197026,
        gdf=state,
        return_col='state',
    )
)
```

```python
# point not in the US
print(
    lookup_geometry_by_point(
        lat=51.500194,
        lng=-115.947875,
        gdf=state,
        return_col='state',
    )
)
```

#### Visualize state data

```python
# Initialize map centered on the continental US
m = folium.Map(location=[38.822591, -100.018268], zoom_start=4.2)

# Add GeoDataFrame layer to map
folium.GeoJson(state['geometry']).add_to(m)

# Define points with state names
points = [
    {'lat': 31.203405, 'lng': -99.592150, 'state': 'Texas'},
    {'lat': 47.605237, 'lng': -122.197026, 'state': 'Washington'},
    # Alberta / outside of the US
    {'lat': 51.500194, 'lng': -115.947875, 'state': 'None'},
]

# Add each point as a red marker with a popup
for point in points:
    folium.Marker(
        location=[point['lat'], point['lng']],
        popup=folium.Popup(point['state'], parse_html=True),
        icon=folium.Icon(color='red'),
    ).add_to(m)

# Display map
m
```
